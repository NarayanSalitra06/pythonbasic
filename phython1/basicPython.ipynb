INDENTATION 

Indentation in the python are use to define the structure and hierarchy of code.  unlike many other programming language that use braces{}  to delimit block of code, python use indentation to determine  the group of statemenet. this mean that all the statement  with in a block must be indented as same level.
 

```python
## indentation
## python use indentation to define a block of code.  consistent use of spaces (commonly 4 spaces)  or the tab is required.

age = 18
if age>15:
    print(age)

print(age)
```

```python
## line comtinuation
## python allow the use of backslash (\) to indicate that a statement continues on the next line.  this is useful for improving the readability of long statement.
total = 1 + 2 + 3 + \
        4 + 5 + 6
print(total)
```

```python
## multiple statement in a single line
x = 10; y = 20; z = x+y
print(z)
```
